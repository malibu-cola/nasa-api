```python
import requests
import os
from dotenv import load_dotenv
from datetime import date, datetime
from IPython.display import Image, display

load_dotenv()
API_KEY = os.getenv("NASA_API_KEY")

today = date.today().isoformat()
url = f"https://api.nasa.gov/EPIC/api/natural/date/2025-04-02"
params = {"api_key": API_KEY}
response = requests.get(url, params=params)
data = response.json()

if not data:
    print("この日付には画像がありません。")
else:
    first_image = data[0]
    dt = datetime.strptime(first_image["date"], "%Y-%m-%d %H:%M:%S")
    image_name = first_image["image"]
    img_url = f"https://epic.gsfc.nasa.gov/archive/natural/{dt.year}/{dt.month:02d}/{dt.day:02d}/png/{image_name}.png"

    print("📅 Date:", first_image["date"])
    print("🛰️ Caption:", first_image["caption"])
    print("🔗 URL:", img_url)
    display(Image(url=img_url))
```
